```python
def group(hitMap):
  groups = [[hitMap]]
  miniGroups = [hitMap]
  while len(miniGroups[0]) >2:
    temp = []
    for i in range(len(miniGroups)):
      middle = len(miniGroups[i])//2
      left = miniGroups[i][0:middle]
      right = miniGroups[i][middle:]
      temp.append(left)
      temp.append(right)
    groups.append(temp)
    miniGroups = temp
  return groups
```

```python
def encode(splitList):
    encodedStr =""
    for i in range (len(splitList)):
      if (len(splitList[i])<=1):
            middle = len(splitList[i][0])//2

            if ('1' in splitList[i][0][0:middle] and'1' in splitList[i][0][middle:] ):
              encodedStr += '11'
            elif ('1' in splitList[i][0:middle]):
              encodedStr += '10'
            else:
              encodedStr += '01'
      else:
        for j in range (len(splitList[i])):
          if (len(splitList[i][j]) >2 ):
            middle = len(splitList[i][j])//2
            if ('1' in splitList[i][j][0:middle] and'1' in splitList[i][j][middle:] ):
              encodedStr += '11'
            elif ('1' in splitList[i][j][0:middle]):
              encodedStr += '10'
            elif('1' in splitList[i][j][middle:]):
              encodedStr += '01'
          else:
            if (splitList[i][j]!="00" ):
              encodedStr += splitList[i][j]

    return encodedStr
```

```python
def bitReplacement(encodedStr):
  #bit replacement
  finalencodedStr = ""
  i = 0;
  while i <(len(encodedStr)):
    if ( encodedStr[i:i+2] == '01' ):
      finalencodedStr+='0'
      #print(finalencodedStr)
      i+=2
    else:
      finalencodedStr+=encodedStr[i:i+2]
      i+=2
  return finalencodedStr
```

```python
def encodeStr(inputString):
  return bitReplacement(encode(group(inputString)))
```
